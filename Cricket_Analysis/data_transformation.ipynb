```python
import pandas as pd
import json
```

**processing match summary**

```python
with open("Json_files/t20_wc_match_results.json") as f:
    data = json.load(f)
    
#data frame creating
df_match=pd.DataFrame(data[0]['matchSummary'])
df_match.head()
```

```python
#checking the dimension of the table
df_match.shape
```

```python
df_match.rename({'scorecard':'match_id'},axis=1,inplace=True)
df_match.head()
```

**creating a match id dictionary that maps team names to unique match id. This will be useful later on to link with other tables**

```python
match_ids_dict={}

for index, row in df_match.iterrows():
    key1 = row['team1']+' Vs '+row['team2']
    key2=row['team2']+' Vs '+row['team1']
    
    match_ids_dict[key1]=row["match_id"]
    match_ids_dict[key2]=row["match_id"]
match_ids_dict
```

```python
df_match.to_csv('csv_files/match_summary.csv',index=False)
```

**processing batting summary**

```python
with open("Json_files/t20_wc_batting_summary.json") as f:
    data = json.load(f)
    all_records=[]
    for rec in data:
        all_records.extend(rec['battingSummary'])

df_batting= pd.DataFrame(all_records)
df_batting.head(25)
```

```python
#creating a new column
df_batting['out/not_out']=df_batting.dismissal.apply(lambda x: "out" if len(x)>0 else "not out")
df_batting.head()
```

```python
df_batting.drop(columns=['dismissal'],inplace=True)
# df_batting.head()
```

```python
df_batting.head(20)
```

```python
df_batting['batsmanName']=df_batting['batsmanName'].apply(lambda x: x.replace('(c)',''))
df_batting['batsmanName']=df_batting['batsmanName'].apply(lambda x: x.replace('â€',''))
df_batting['batsmanName']=df_batting['batsmanName'].apply(lambda x: x.replace('\xa0',''))
df_batting.head(20)
```

```python
#creating a new column in batting df to join with match summary table
df_batting['match_id']=df_batting["match"].map(match_ids_dict)
df_batting.head()
```

```python
df_batting.shape
```

```python
df_batting.to_csv('csv_files/batting_summary.csv',index=False)
```

**Processing bowling summary**

```python
with open('Json_files/t20_wc_bowling_summary.json') as f:
    data = json.load(f)
    all_records = []
    for rec in data:
        all_records.extend(rec['bowlingSummary'])
all_records[:2]
```

```python
df_bowling = pd.DataFrame(all_records)
print(df_bowling.shape)
df_bowling.head()
```

```python
df_bowling['match_id'] = df_bowling['match'].map(match_ids_dict)
df_bowling.head()
```

```python
df_bowling.to_csv('csv_files/bowling_summary.csv', index = False)
```

**Proccessing Player information**
