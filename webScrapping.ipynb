```python
#connect to the website
url = 'http://books.toscrape.com/'
# You can add headers if needed, but for this site, it's usually not required
headers={"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/139.0.0.0 Safari/537.36 Edg/139.0.0.0"}

page = requests.get(url) # Removed headers as they are not necessary for this site
soup1 = BeautifulSoup(page.content, "html.parser")
# For this site, prettify() is not strictly necessary to find the title, but you can use it
soup2 = BeautifulSoup(soup1.prettify(), "html.parser")

# Find the title of the first book
# The book titles are within <h3> tags, inside a product_pod div
title = soup1.find('article', class_='product_pod').h3.a['title'] # Used soup1 as prettify() is optional

# Find the price of the first book
# The price is within a <p class="price_color"> tag
price = soup1.find('article', class_='product_pod').find('p', class_='price_color').get_text()
print(title)
price=price.strip()[1:]
print(price)
```

```python
import datetime
today=datetime.date.today()
print(today)
```

```python
#create a csv then import and automate the data automatically
import csv
header = ['Title', 'Price','Date']
data = [title, price,today]
# The filename needs to be in quotes
with open('scrappingData.csv','w', newline='', encoding='UTF8') as f:
    writer = csv.writer(f) #creating the csv
    writer.writerow(header) #inserting the header
    writer.writerow(data) #inserting the data
```

```python
#automating the data entry into the csv
def check_price():
  #connect to the website
  url = 'http://books.toscrape.com/'
  headers={"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/139.0.0.0 Safari/537.36 Edg/139.0.0.0"}
  page = requests.get(url)
  soup1 = BeautifulSoup(page.content, "html.parser")
  soup2 = BeautifulSoup(soup1.prettify(), "html.parser")

  title = soup1.find('article', class_='product_pod').h3.a['title']
  price = soup1.find('article', class_='product_pod').find('p', class_='price_color').get_text()
  price=price.strip()[1:]
  import datetime
  today=datetime.date.today()
  import csv
  header = ['Title', 'Price','Date']
  data = [title, price,today]
  with open('scrappingData.csv','a+', newline='', encoding='UTF8') as f:
    writer = csv.writer(f) #creating the csv
    writer.writerow(data) #inserting the data
```
